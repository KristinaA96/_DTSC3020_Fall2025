```python
import requests

HEADERS = {"User-Agent": (
    "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
    "(KHTML, like Gecko) Chrome/122.0 Safari/537.36")}

def fetch_html(url: str, timeout: int = 20) -> str:
    r = requests.get(url, headers=HEADERS, timeout=timeout)
    r.raise_for_status()
    return r.text

URL_Q1 = "https://www.iban.com/country-codes"
html_content_q1 = fetch_html(URL_Q1)
print("HTML content fetched for Question 1.")
```

```python
import pandas as pd

def flatten_headers(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [" ".join([str(x) for x in tup if str(x)!="nan"]).strip()
                      for tup in df.columns.values]
    else:
        df.columns = [str(c).strip() for c in df.columns]
    return df

def q1_read_table(html: str) -> pd.DataFrame:
    dfs = pd.read_html(html)
    for df in dfs:
        if df.shape[1] >= 3:
            return flatten_headers(df)
    raise ValueError("No suitable table found with at least 3 columns.")

df_q1 = q1_read_table(html_content_q1)
print("Table extracted into DataFrame.")
df_q1.head()
```

```python
import pandas as pd

def q1_clean(df: pd.DataFrame) -> pd.DataFrame:
    """Clean columns: strip, UPPER Alpha-2/Alpha-3, cast Numeric to int (nullable), drop invalids.
    """
    df_cleaned = df.copy()

    # Trim spaces from all string columns
    for col in df_cleaned.select_dtypes(include=['object']).columns:
        df_cleaned[col] = df_cleaned[col].astype(str).str.strip()

    # Convert Alpha-2 and Alpha-3 to UPPERCASE
    for col_name in ['Alpha-2 code', 'Alpha-3 code']:
        if col_name in df_cleaned.columns:
            df_cleaned[col_name] = df_cleaned[col_name].astype(str).str.upper()

    # Cast Numeric to nullable integer
    if 'Numeric' in df_cleaned.columns:
        df_cleaned['Numeric'] = pd.to_numeric(df_cleaned['Numeric'], errors='coerce').astype(pd.Int64Dtype())

    return df_cleaned

df_q1_cleaned = q1_clean(df_q1)
print("DataFrame cleaned.")
df_q1_cleaned.head()
```

```python
def q1_sort_top(df: pd.DataFrame, top: int = 15) -> pd.DataFrame:
    """Sort descending by Numeric and return Top-N.
    """
    # Ensure 'Numeric' column exists and is sortable
    if 'Numeric' not in df.columns:
        raise ValueError("DataFrame must contain a 'Numeric' column.")

    # Sort by 'Numeric' descending, handling NaNs if any (they will go to the end)
    sorted_df = df.sort_values(by='Numeric', ascending=False, na_position='last')

    return sorted_df.head(top)


df_q1_top_15 = q1_sort_top(df_q1_cleaned)
print("Top 15 rows sorted by Numeric (descending):")
display(df_q1_top_15)
```

```python
import pandas as pd
from bs4 import BeautifulSoup

# --- Q2 Skeleton (fill the TODOs) ---
def q2_parse_items(html: str) -> pd.DataFrame:
    """Parse front page items into DataFrame columns:
       rank, title, link, points, comments, user (optional).
    TODO: implement with BeautifulSoup on '.athing' and its sibling '.subtext'.
    """
    soup = BeautifulSoup(html, 'lxml')
    items_data = []

    stories = soup.find_all('tr', class_='athing')

    for story in stories:
        rank_tag = story.find('span', class_='rank')
        rank = rank_tag.text.replace('.', '') if rank_tag else None

        title_link_tag = story.find('span', class_='titleline')
        title = title_link_tag.find('a').text if title_link_tag and title_link_tag.find('a') else None
        link = title_link_tag.find('a')['href'] if title_link_tag and title_link_tag.find('a') else None

        # Details (points, comments, user) are in the next tr with .subtext
        subtext_row = story.find_next_sibling('tr')
        subtext_td = subtext_row.find('td', class_='subtext') if subtext_row else None

        points = None
        comments = None
        user = None

        if subtext_td:
            score_span = subtext_td.find('span', class_='score')
            if score_span:
                points_text = score_span.text.split(' ')[0]
                if points_text.isdigit():
                    points = int(points_text)

            comments_link = subtext_td.find('a', string=lambda text: text and 'comments' in text)
            if comments_link:
                # Extract numeric part from 'X comments' or 'discuss'
                comments_text = comments_link.text
                if 'comments' in comments_text:
                    try:
                        comments = int(comments_text.split(' ')[0])
                    except ValueError:
                        comments = 0 # Handle 'discuss' or other non-numeric cases
                elif 'discuss' in comments_text:
                    comments = 0 # No comments yet

            user_link = subtext_td.find('a', class_='hnuser')
            user = user_link.text if user_link else None

        items_data.append({
            'rank': rank,
            'title': title,
            'link': link,
            'points': points,
            'comments': comments,
            'user': user
        })

    # Filter out potential None rows if any parsing failed completely for a 'story' item
    items_data = [item for item in items_data if any(value is not None for value in item.values())]

    return pd.DataFrame(items_data)

def q2_clean(df: pd.DataFrame) -> pd.DataFrame:
    """Clean numeric fields and fill missing values.
    TODO: cast points/comments/rank to int (non-digits -> 0). Fill text fields.
    """
    raise NotImplementedError("TODO: implement q2_clean")

def q2_sort_top(df: pd.DataFrame, top: int = 15) -> pd.DataFrame:
    """Sort by points desc and return Top-N. TODO: implement."""
    raise NotImplementedError("TODO: implement q2_sort_top")
```

```python
import re, sys, pandas as pd, requests
from bs4 import BeautifulSoup
HEADERS = {"User-Agent": (
    "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
    "(KHTML, like Gecko) Chrome/122.0 Safari/537.36")}
def fetch_html(url: str, timeout: int = 20) -> str:
    r = requests.get(url, headers=HEADERS, timeout=timeout)
    r.raise_for_status()
    return r.text
def flatten_headers(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [" ".join([str(x) for x in tup if str(x)!="nan"]).strip()
                      for tup in df.columns.values]
    else:
        df.columns = [str(c).strip() for c in df.columns]
    return df
print("Common helpers loaded.")
```

```python
import pandas as pd

def q2_clean(df: pd.DataFrame) -> pd.DataFrame:
    """Clean numeric fields and fill missing values.
    Cast points/comments/rank to int (non-digits -> 0). Fill text fields.
    """
    df_cleaned = df.copy()

    # Cast numeric columns to int, coercing errors to 0
    for col in ['rank', 'points', 'comments']:
        if col in df_cleaned.columns:
            # Ensure 'rank' is treated as numeric before casting to int
            # For points/comments, we already tried to cast to int during parsing, but handle NaNs/None
            df_cleaned[col] = pd.to_numeric(df_cleaned[col], errors='coerce').fillna(0).astype(int)

    # Fill missing text fields with an empty string or 'N/A'
    for col in ['title', 'link', 'user']:
        if col in df_cleaned.columns:
            df_cleaned[col] = df_cleaned[col].fillna('') # Or fill with 'N/A'

    return df_cleaned

df_q2_cleaned = q2_clean(df_q2)
print("Hacker News DataFrame cleaned.")
df_q2_cleaned.head()
```

```python
import pandas as pd

def q2_sort_top(df: pd.DataFrame, top: int = 15) -> pd.DataFrame:
    """Sort by points desc and return Top-N."""
    if 'points' not in df.columns:
        raise ValueError("DataFrame must contain a 'points' column.")

    # Sort by 'points' descending. NaNs (if any, though cleaned to 0) would go to the end.
    sorted_df = df.sort_values(by='points', ascending=False)

    return sorted_df.head(top)

df_q2_top_15 = q2_sort_top(df_q2_cleaned)
print("Top 15 Hacker News stories sorted by points (descending):")
display(df_q2_top_15)
```
