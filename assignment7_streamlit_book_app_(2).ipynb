```python
# Write your code for Part 1 here

import streamlit as st
import pandas as pd

# Set a page title for the app
st.set_page_config(page_title="My Book Library App")

# Create a list of dictionaries with at least 5 books
books_data = [
    {"Title": "The Hitchhiker's Guide to the Galaxy", "Author": "Douglas Adams", "Genre": "Sci-Fi", "Pages": 208, "Read": True},
    {"Title": "Pride and Prejudice", "Author": "Jane Austen", "Genre": "Classic", "Pages": 279, "Read": False},
    {"Title": "1984", "Author": "George Orwell", "Genre": "Dystopian", "Pages": 328, "Read": True},
    {"Title": "To Kill a Mockingbird", "Author": "Harper Lee", "Genre": "Fiction", "Pages": 281, "Read": True},
    {"Title": "The Great Gatsby", "Author": "F. Scott Fitzgerald", "Genre": "Classic", "Pages": 180, "Read": False},
    {"Title": "Dune", "Author": "Frank Herbert", "Genre": "Sci-Fi", "Pages": 412, "Read": False}
]

# Convert the list to a pandas DataFrame
books_df = pd.DataFrame(books_data)

# Use st.session_state to keep the data after adding new books
if "books_df" not in st.session_state:
    st.session_state["books_df"] = books_df
```


## Part 3 – Add New Book Form (1.0 point)

**Requirements:**

- Create a small form that allows the user to add a new book.
- The form must include inputs for:
  - `Title`  (text input)
  - `Author` (text input)
  - `Genre`  (selectbox or text input)
  - `Pages`  (number input, must be > 0)
  - `Read`   (checkbox)
- When the user submits the form:
  - **Validate** the inputs:
    - `Title` and `Author` cannot be empty.
    - `Pages` must be a positive number.
  - If validation passes:
    - Append the new book to the DataFrame.
    - Show a **success message**.
    - Update the displayed table.
  - If validation fails:
    - Show an **error message**.

**Hints:**
- You can use `st.form()` and `st.form_submit_button()` to create the form.
- To update the table, you must update the same DataFrame you used in Part 2.

>  Write your code for **Part 3** in the next cell.


```python
# Write your code for Part 3 here

import streamlit as st
import pandas as pd

# Retrieve the DataFrame from session_state
books_df = st.session_state["books_df"]

# Create a subheader for the form
st.subheader("Add a New Book")

# Create a form with title, author, genre, pages, read
with st.form("add_book_form"):
    title = st.text_input("Title")
    author = st.text_input("Author")

    # Example of using st.selectbox for Genre
    genre_options = ["Fiction", "Non-Fiction", "Sci-Fi", "Fantasy", "Biography", "Classic", "Dystopian", "Mystery", "Thriller", "Horror", "Romance", "History", "Science"]
    genre = st.selectbox("Genre", options=genre_options)

    pages = st.number_input("Pages", min_value=1, step=1)
    read = st.checkbox("Read")

    submitted = st.form_submit_button("Add Book")

    if submitted:
        # Validation logic
        if not title or not author:
            st.error("Title and Author cannot be empty.")
        elif pages <= 0:
            st.error("Pages must be a positive number.")
        else:
            # Create a new book entry
            new_book_data = {
                "Title": title,
                "Author": author,
                "Genre": genre,
                "Pages": pages,
                "Read": read
            }
            new_book_df = pd.DataFrame([new_book_data])

            # Append the new row to the DataFrame and update session_state
            st.session_state["books_df"] = pd.concat([st.session_state["books_df"], new_book_df], ignore_index=True)
            st.success("New book added successfully!")

            # Re-display the updated DataFrame so the user can see the new book
            st.experimental_rerun() # This forces Streamlit to rerun the entire script to show the updated table
```
